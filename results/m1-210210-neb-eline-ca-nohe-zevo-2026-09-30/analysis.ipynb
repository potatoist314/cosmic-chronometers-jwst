# M1_210210: production-defaults base with and without metallicity evolution

- neb_eline_ca_nohe: nebular grid; free line fluxes, no H$\epsilon$ column; Ca H+K, Ca I 4227 masked; wide dust priors.
- neb_eline_ca_nohe_zevo: as neb_eline_ca_nohe with `metallicity_evolution` on.
- Same data, seed and SFH bins.
- zevo $Z(m) = Z_f - (Z_f - Z_0)(1 - m)^\alpha$, $m$ formed-mass fraction; $\beta = 1/(1+\alpha)$ from `zh_beta_unit`; $Z_0 = \min(Z_\odot/50, \langle Z\rangle)$.
- zevo `Z` is $\log_{10}\langle Z\rangle$, formed-mass weighted.
- Metallicity is $[\mathrm{Fe}/\mathrm{H}]$ = grid $Z$ + `FEH_OFFSET`.

```python
import os
import re
import sys
from pathlib import Path

os.environ.setdefault("JAX_PLATFORMS", "cpu")
import corner
import h5py
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

PROJECT_ROOT = Path(os.environ.get("CERIDWEN_PROJECT_ROOT", Path.cwd()))
while not (PROJECT_ROOT / "scripts" / "per_galaxy_diagnostics.py").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT / "scripts"))
import per_galaxy_diagnostics as pgd  # noqa: E402
from ceridwen.model import mass_mapped_beta  # noqa: E402
from build_dr2_quiescent_summary import FEH_OFFSET, formation_times  # noqa: E402
from spectral_figures import (  # noqa: E402
    mark_absorption_features,
    mark_rest_wavelength_axis,
    set_plain_log_ticks,
    spectral_tight_layout,
)

RESULTS = PROJECT_ROOT / "results/m1-210210-neb-eline-ca-nohe-zevo-2026-09-30"
TARGET = "M1_210210"
FIT_DIRS = {"neb_eline_ca_nohe": PROJECT_ROOT / "results/m1-210210-neb-eline-ca-nohe-2026-09-30/run/fits/neb_eline_ca_nohe" / f"210210-{TARGET}",
            "neb_eline_ca_nohe_zevo": RESULTS / "run/fits/neb_eline_ca_nohe_zevo" / f"210210-{TARGET}"}
COLOURS = {"neb_eline_ca_nohe": "#222222", "neb_eline_ca_nohe_zevo": "#0072B2"}
NAMES = {"neb_eline_ca_nohe": "neb_eline_ca_nohe: constant Z",
         "neb_eline_ca_nohe_zevo": r"neb_eline_ca_nohe_zevo: $Z(m)$"}
REF = "neb_eline_ca_nohe"
LOGZ_MAX = float(h5py.File(Path("~/.ceridwen/grids/amist_c3k_hr_krou_afe_nebular.h5").expanduser(), "r")["ssp_lgmet"][:].max())
LOGZ_0_CAP = -FEH_OFFSET - np.log10(50.0)  # grid Z at [Fe/H] = log10(1/50), as in the production notebook
LABELS = {"logmass": r"$\log_{10}(M_\star/M_\odot)$", "Z": r"$[\mathrm{Fe}/\mathrm{H}]$", "afe": r"$[\alpha/\mathrm{Fe}]$",
          "tau_dust": r"$\tau_{\mathrm{dust}}$", "dust_index": r"$\delta_{\rm dust}$", "ebump": r"$E_{\rm bump}$", "dust_ratio": r"$r_{\mathrm{dust}}$",
          "beta": r"$\beta_{\rm ZH}$", "age": r"$t_{\mathrm{MW}}$ [Gyr]"}
pd.set_option("display.width", 250); pd.set_option("display.max_columns", 60); pd.set_option("display.precision", 3)


def dust_samples(g):
    """tau_dust, delta_dust, E_bump; the Kriek & Conroy bump is tied, 0.85 - 1.9 delta (sedpy_jax kriek_conroy)."""
    s = g.samples
    if "diffuse_tau_noll" in s:
        return np.ravel(s["diffuse_tau_noll"]), np.ravel(s["diffuse_delta"]), np.ravel(s["diffuse_Ebump"])
    delta = np.ravel(s["diffuse_dust_index"])
    return np.ravel(s["diffuse_tau_kc"]), delta, 0.85 - 1.9 * delta


def beta_samples(g):
    """Physical ZH slope beta from zh_beta_unit; NaN when the fit has constant Z."""
    if "zh_beta_unit" not in g.samples:
        return np.full(len(np.ravel(g.samples["logmass"])), np.nan)
    logz = np.ravel(g.samples["Z"]); logz_0 = np.minimum(logz, LOGZ_0_CAP)
    return np.asarray(mass_mapped_beta(np.ravel(g.samples["zh_beta_unit"]), logz, logz_0, LOGZ_MAX))
```

## Parameters, evidence and $\chi^2$

- `spec_chi2_raw`: catalogue uncertainties only, posterior-median model.
- `spec_chi2_fref`: every fit at the neb_eline_ca_nohe fit's calibration floor.
- `phot_chi2` and `phot_n` refer to each fit's own bands.
- `u_pull`: CFHT $u^*$ (data − model) / uncertainty at the posterior median.
- `*_q16`, `*_q50`, `*_q84`: weighted 16th, 50th, 84th percentiles.
- `beta`, `Z_0`, `Z_f`: zevo enrichment shape and initial, final $[\mathrm{Fe}/\mathrm{H}]$.
- `mf_<a>-<b>`: formed-mass fraction in the bin from $a$ to $b$ Gyr.

```python
def _text(v):
    return v.decode() if isinstance(v, bytes) else str(v)


def load_fit(folder):
    out = {"folder": folder}
    with h5py.File(folder / "ceridwen_derived_outputs.h5", "r") as d:
        out["attrs"] = dict(d.attrs)
        names = [_text(v) for v in d["summary/parameter"][:]]
        out["summary"] = pd.DataFrame({"q16": d["summary/q16"][:], "q50": d["summary/q50"][:], "q84": d["summary/q84"][:]}, index=names)
        out["diag"] = dict(d["diagnostics"].attrs)
        for grp in ("spectrum", "photometry", "sfh", "calibration"):
            out[grp] = {k: d[grp][k][:] for k in d[grp]}
        out["calibration"]["attrs"] = dict(d["calibration"].attrs)
    with h5py.File(folder / "ceridwen_result.h5", "r") as r:
        out["tau_max"] = float(re.search(r"diffuse_tau_(?:kc|noll): .*?Uniform\(0, ([0-9.]+)\)", _text(r["model"].attrs["parameter_block"])).group(1))
        out["n_calls"] = int(r["samples"].attrs["n_likelihood_calls"])
        out["wall_s"] = float(r["samples"].attrs["wall_time_s"])
    out["galaxy"] = pgd.load_galaxy(folder)
    return out


fits = {k: load_fit(v) for k, v in FIT_DIRS.items()}
REF_FLOOR = fits[REF]["summary"].loc["calibration floor [%]", "q50"]


def chi2_at_floor(fit, f_pct):
    spec = fit["spectrum"]; mask = spec["mask"].astype(bool)
    sigma = np.hypot(spec["uncertainty"], (f_pct / 100.0) * np.abs(spec["posterior_q50"]))
    return float(np.sum(((spec["observed"] - spec["posterior_q50"]) / sigma)[mask] ** 2))


def summary_row(fit):
    s, d = fit["summary"], fit["diag"]
    g = fit["galaxy"]; w = pgd.posterior_weights(g)
    t20, t50, t80 = formation_times(fit["sfh"]["lookback_time_gyr"], fit["sfh"]["mass_fraction_draws"])
    row = dict(order=int(fit["calibration"]["attrs"]["order"]), tau_max=fit["tau_max"], seed=g.seed, lnZ=float(d["log_evidence"]), lnZ_err=float(d["log_evidence_err"]),
               n_pix=int(fit["spectrum"]["mask"].sum()), spec_chi2_raw=chi2_at_floor(fit, 0.0), spec_chi2_fref=chi2_at_floor(fit, REF_FLOOR),
               spec_chi2_stored=float(d["spectrum_chi2"]), phot_chi2=float(d["photometry_chi2"]), phot_n=int(d["photometry_ndof"]),
               ess=float(d["posterior_weight_ess"]), passed=bool(d["passed"]), calls=fit["n_calls"], wall_s=fit["wall_s"])
    for name, key in [("logmass", "logmass"), ("Z", "Z"), ("afe", "afe"),
                      ("age", "mass-weighted age [Gyr]"), ("f_calib", "calibration floor [%]")]:
        for q in ("q16", "q50", "q84"):
            row[f"{name}_{q}"] = s.loc[key, q] + (FEH_OFFSET if name == "Z" else 0.0)
    for name, x in zip(("tau_dust", "dust_index", "ebump"), dust_samples(g)):
        row[f"{name}_q16"], row[f"{name}_q50"], row[f"{name}_q84"] = (pgd.weighted_quantile(x, w, q) for q in (0.16, 0.5, 0.84))
    for name, t in [("t20", t20), ("t50", t50), ("t80", t80)]:
        row[f"{name}_q16"], row[f"{name}_q50"], row[f"{name}_q84"] = np.percentile(t, [16, 50, 84])
    phot = fit["photometry"]
    row["u_pull"] = float(phot["pull"][[_text(f) for f in phot["filters"]].index("cfht_megacam_us_9301")])
    if "dust_ratio" in g.samples:
        row["dust_ratio_q16"], row["dust_ratio_q50"], row["dust_ratio_q84"] = (
            pgd.weighted_quantile(np.ravel(g.samples["dust_ratio"]), w, q) for q in (0.16, 0.5, 0.84))
    if "zh_beta_unit" in g.samples:  # Z_f = Z_0 + (<Z> - Z_0) / (1 - beta), linear Z
        logz = np.ravel(g.samples["Z"]); logz_0 = np.minimum(logz, LOGZ_0_CAP)
        beta = beta_samples(g)
        logz_f = np.log10(10.0 ** logz_0 + (10.0 ** logz - 10.0 ** logz_0) / (1.0 - beta))
        for name, x in (("beta", beta), ("Z_0", logz_0 + FEH_OFFSET), ("Z_f", logz_f + FEH_OFFSET)):
            row[f"{name}_q16"], row[f"{name}_q50"], row[f"{name}_q84"] = (pgd.weighted_quantile(x, w, q) for q in (0.16, 0.5, 0.84))
    edges = fit["sfh"]["lookback_time_gyr"]
    for i in range(len(edges) - 1):
        for q in ("q16", "q50", "q84"):
            row[f"mf_{edges[i]:.2f}-{edges[i + 1]:.2f}_{q}"] = fit["sfh"][f"mass_fraction_{q}"][i]
    for key in ("zred", "sigma_smooth"):
        if key in s.index:
            for q in ("q16", "q50", "q84"):
                row[f"{key}_{q}"] = s.loc[key, q]
    return row


table = pd.DataFrame({k: summary_row(v) for k, v in fits.items()}).T
table.to_csv(RESULTS / "comparison.csv")
display(table.T)
```

## SFH and corner

- SFH median with 16-84 band; corner uses weighted dead points, 1σ contours.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 3.4))
for k, fit in fits.items():
    sfh = fit["sfh"]; lb = sfh["lookback_time_gyr"]
    axes[0].plot(lb, sfh["normalized_sfr_q50"], color=COLOURS[k], lw=1.4, label=NAMES[k])
    axes[0].fill_between(lb, sfh["normalized_sfr_q16"], sfh["normalized_sfr_q84"], color=COLOURS[k], alpha=0.12, lw=0)
    younger = np.concatenate([[0.0], np.cumsum(sfh["mass_fraction_q50"])])
    axes[1].plot(lb, younger / younger[-1], color=COLOURS[k], lw=1.4, label=NAMES[k])
axes[0].set(xscale="log", yscale="log", xlabel="lookback time [Gyr]", ylabel=r"SFR / $M_{\rm formed}$ [yr$^{-1}$]", xlim=(0.02, lb[-1] * 1.05))
axes[1].set(xlabel="lookback time [Gyr]", ylabel="mass fraction younger", ylim=(0, 1.02))
axes[0].legend(frameon=False, fontsize=8)
fig.tight_layout(); fig.savefig(RESULTS / f"sfh-{TARGET}.png", dpi=130); plt.show()
```

```python
CORNER_PARAMS = ["logmass", "Z", "afe", "tau_dust", "dust_index", "ebump", "dust_ratio", "beta", "age"]


def mw_age_samples(g):
    r = np.asarray(g.samples["logsfr_ratios"])
    h = 10.0 ** np.concatenate([np.zeros((len(r), 1)), -np.cumsum(r, axis=1)], axis=1)
    edges = g.sfh_edges_gyr
    m = 0.5 * (h[:, :-1] + h[:, 1:]) * np.diff(edges)
    return (m * (0.5 * (edges[:-1] + edges[1:]))).sum(1) / m.sum(1)


def corner_data(g):
    return np.column_stack([np.ravel(g.samples["logmass"]), np.ravel(g.samples["Z"]) + FEH_OFFSET, np.ravel(g.samples["afe"]), *dust_samples(g),
                            np.ravel(g.samples["dust_ratio"]), beta_samples(g), mw_age_samples(g)])


data = {k: (corner_data(f["galaxy"]), pgd.posterior_weights(f["galaxy"])) for k, f in fits.items()}
n = len(CORNER_PARAMS)
lo, hi = np.full(n, np.inf), np.full(n, -np.inf)
for x, w in data.values():  # beta is finite only in the zevo arm; range each panel over finite draws
    for i in range(n):
        finite = np.isfinite(x[:, i])
        if finite.sum() > 10:
            lo[i] = min(lo[i], pgd.weighted_quantile(x[finite, i], w[finite], 0.002))
            hi[i] = max(hi[i], pgd.weighted_quantile(x[finite, i], w[finite], 0.998))
rng_ = [(a, b) if b > a else (a - 1e-3, a + 1e-3) for a, b in zip(lo, hi)]
zevo_key = next(k for k in fits if "zh_beta_unit" in fits[k]["galaxy"].samples)
x0, w0 = data[zevo_key]
fig = corner.corner(x0, weights=w0, labels=[LABELS[p] for p in CORNER_PARAMS], range=rng_, color=COLOURS[zevo_key], bins=40, smooth=1.0,
                    plot_datapoints=False, plot_density=False, fill_contours=False, levels=[1 - np.exp(-0.5)],
                    hist_kwargs=dict(density=True, lw=1.4), contour_kwargs=dict(linewidths=1.4))
for k, (x, w) in data.items():  # constant-Z arms overlay the shared panels only, same bins and levels
    if k == zevo_key:
        continue
    for i, p in enumerate(CORNER_PARAMS):
        if p == "beta" or not np.isfinite(x[:, i]).any():
            continue
        fig.axes[i * (n + 1)].hist(x[:, i], bins=np.linspace(*rng_[i], 41), weights=w, histtype="step", density=True, color=COLOURS[k], lw=1.4)
        for j in range(i):
            if CORNER_PARAMS[j] == "beta":
                continue
            corner.hist2d(x[:, j], x[:, i], weights=w, ax=fig.axes[i * n + j], range=[rng_[j], rng_[i]], bins=40, smooth=1.0,
                          color=COLOURS[k], plot_datapoints=False, plot_density=False, fill_contours=False, levels=[1 - np.exp(-0.5)],
                          contour_kwargs=dict(linewidths=1.4))
for i in range(n):  # later corner calls reset each diagonal's y-limit; fit the tallest arm
    ax = fig.axes[i * (n + 1)]
    ax.set_ylim(0, 1.05 * max(patch.get_path().vertices[:, 1].max() for patch in ax.patches))
fig.legend(handles=[plt.Line2D([], [], color=COLOURS[k], label=NAMES[k]) for k in fits], loc="upper right", frameon=False, fontsize=18)
fig.suptitle(TARGET, y=0.98)
fig.savefig(RESULTS / f"corner-{TARGET}.png", dpi=150); plt.show()
```
